```python
from glob import glob
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Data Analysis_V2_Combined_JPD_Ml_v4.csv", index_col=0)
```

```python
display(df.head(24))
```

```python
# import pycaret classification and init setup
from pycaret.classification import *
setup(df, target = 'Carcinoma Status', train_size = 0.8, session_id = 123)
```

```python
# compare baseline models
best = compare_models()
print(best)
```

```python
# create model
model = create_model('lda')
```

```python
# print model parameters
print(model)
```

```python
# tune hyperparameters of et
tuned_model = tune_model(model)
```

```python
# plot feature importance
plot_model(tuned_model, plot = 'feature',save=True)
```

```python
# Plot AUC
plot_model(tuned_model, plot='auc', save=True)
```
